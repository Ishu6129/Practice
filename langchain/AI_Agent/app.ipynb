```python
from langchain_groq import ChatGroq
from langchain.tools import tool
import requests
from dotenv import load_dotenv
load_dotenv() 
```

```python
from langchain_community.tools import DuckDuckGoSearchRun

search_tool=DuckDuckGoSearchRun()
result=search_tool.invoke("Raam mandir latest news")
```

```python
result
```

```python
@tool
def get_weather_data(city: str) -> str:
  """
  This function fetches the current weather data for a given city
  """
  url = f'https://api.weatherstack.com/current?access_key=4d1d8ae207a8c845a52df8a67bf3623e&query={city}'

  response = requests.get(url)

  return response.json()
```

```python
llm=ChatGroq(model="llama-3.3-70b-versatile")
llm.invoke("tell me your model name")
```

```python
from langchain_classic.agents import AgentExecutor,create_react_agent
```

```python
from langsmith import Client

client = Client()

prompt = client.pull_prompt(
    "hwchase17/react",
    dangerously_pull_public_prompt=True  # ONLY USE TO FETCH PROMPT FROM TRUSTED SOURCE 
)
```

```python
prompt.template
```

```python
# Agent plans,break down tasks and decide which tool to use
agent=create_react_agent(
    llm=llm,
    tools=[search_tool,get_weather_data],
    prompt=prompt
)
```

```python
# Follow the steps give by agent and execute it
agent_executor=AgentExecutor(
    verbose=True,
    agent=agent,
    tools=[search_tool,get_weather_data]
)
```

```python
response=agent_executor.invoke({"input":"What is the capital of India and what is the weather of it's capital today and result should be in markdown"})
print(response)
```
