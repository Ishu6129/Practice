# GET ALL AVAILABLE MODELS LIST

```python
from google import genai
from dotenv import load_dotenv
load_dotenv()
models = genai.Client().models.list()
for model in models:
    print(model.name)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
import textwrap
from dotenv import load_dotenv

load_dotenv()

model = ChatGoogleGenerativeAI(model='gemini-2.5-flash',temperature=1.5,max_output_tokens=30)

result = model.invoke('What is the capital of India')
print(textwrap.fill(str(result), width=80))
print("-----METADATA-----\n",result.usage_metadata)
print("-----------------\n",result.content)
```
